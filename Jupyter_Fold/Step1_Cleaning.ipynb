```python
for i in df.index:
    DurCol = df.loc[i,'Duration']
    if " " in DurCol:
        column1 = DurCol.split(" ")[0]
        column2 = DurCol.split(" ")[1]
        
        if "h" in column1:
            column1= int(column1.replace("h",""))*60
        elif "m" in column1:
            column1= int(column1.replace("m",""))
        
        if "h" in column2:
            column2 = int(column2.replace("h",""))*60
        elif "m" in column2:
            column2= int(column2.replace("m",""))
        print(column2)
    
        df.loc[i,'DurHrs_Mins'] = column1 + column2
        
    else:
        if "h" in DurCol:
            DurCol = int(DurCol.replace("h",""))*60
            df.loc[i,'DurHrs_Mins'] = DurCol
        elif "m" in DurCol:
            DurCol = int(DurCol.replace("m",""))
            df.loc[i,'DurHrs_Mins'] = DurCol

```
