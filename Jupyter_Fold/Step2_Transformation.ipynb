```python
import pandas as pd

X_train = pd.read_csv(r"C:\Users\SWATI\Desktop\Internship Projects\Final_Project\Airline_ML_01\Data\Cleaned_Data\X_train.csv")
X_test = pd.read_csv(r"C:\Users\SWATI\Desktop\Internship Projects\Final_Project\Airline_ML_01\Data\Cleaned_Data\X_test.csv")
```

```python
X_train
```

```python

# Transformation encoding
# Ordinal Col
# one hote col
# Numerical Col

NumericalCol = ['Total_Stops','DurHrs_Mins','Dept_Day','Dept_Month','Dept_Year','Dept_Hour','Dept_Minute','Arv_Hour','Arv_Minute','Arv_Day','Arv_Month','Arv_Year']
OneHotCol =['Airline','Source','Destination'] 
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

OneHotEncoding_01 =  ("One_Hot_Encoder",OneHotEncoder(sparse_output=False, drop="first", handle_unknown="ignore"),OneHotCol,)
Scaling = ("Numerical_Coluns", StandardScaler(), NumericalCol)

transform_Obj = ColumnTransformer(transformers=[OneHotEncoding_01,Scaling]).set_output(transform="pandas")

transform_Obj
```

```python
X_train_trans = transform_Obj.fit_transform(X_train)
X_train_trans.to_csv(r"C:\Users\SWATI\Desktop\Internship Projects\Final_Project\Airline_ML_01\Data\Trans_Data/X_Train_Trans.csv")
```

```python
import joblib
joblib.dump(transform_Obj,r"C:\Users\SWATI\Desktop\Internship Projects\Final_Project\Airline_ML_01\TransFile/Transform.pkl")
```
